```python
import string

def remove_punctuation(str):

    cleaned_text = ""

    for char in str:
        if char not in string.punctuation:
            cleaned_text += char

    return cleaned_text        

sample_review = dataset["train"][0]["text"]

print("Original Review:")
print(sample_review[:300])

print("\nCleaned Review:")
print(remove_punctuation(sample_review)[:300])
```

```python
from nltk.corpus import stopwords

stop_words = stopwords.words('english')

def remove_stopwords(text):
    
    words = text.split()

    filtered = []

    for word in words:
        if word.lower() not in stop_words:
            filtered.append(word)

    filtered_text = " ".join(filtered)
    return filtered_text

print(remove_stopwords("This movie is really amazing and I love it"))



```

```python
def preprocess_text(text):

    text = text.lower()

    text = remove_punctuation(text)
    
    text = remove_stopwords(text)

    return text
    
```

```python
text = "This MOVIE was AMAZING!!! I really loved it."

print(preprocess_text(text))
```

```python
reviews = dataset["train"]["text"]

reviews = reviews[:20]

```

```python
def build_vocab(reviews):

    vocab = set()

    for text in reviews:
        words = text.split()
        for word in words:
            vocab.add(word) 


    return vocab          

```

```python
vocab = build_vocab(reviews)
print(len(vocab)) #No. of unique words in all the reviews
```

```python
def review_to_vector(review, vocab):

    vector = []

    words = review.split()

    for word in vocab:
        
            vector.append(words.count(word))
      

    return vector
 
review = "good movie good love"

print(review_to_vector(review, vocab))    
                
```

```python
def reviews_to_matrix(reviews, vocab):
    matrix = []

    for review in reviews:

        matrix.append(review_to_vector(review,vocab))

    return matrix

reviews = [
    "good movie",
    "bad movie",
    "good love movie"
]

vocab = ['good', 'bad', 'movie', 'love']

print(reviews_to_matrix(reviews, vocab))    
        
```

```python
from sklearn.feature_extraction.text import CountVectorizer

reviews = [
    "good movie",
    "bad movie",
    "good love movie"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(reviews)

print(X.toarray())
```
